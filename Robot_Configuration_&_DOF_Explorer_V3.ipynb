```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

L1 = 4
L2 = 3
L3 = 2

joint1 = widgets.IntSlider(
    value=45,
    min=-180,
    max=180,
    description='Joint1'
)

joint2 = widgets.IntSlider(
    value=30,
    min=-180,
    max=180,
    description='Joint2'
)

joint3 = widgets.IntSlider(
    value=60,
    min=-180,
    max=180,
    description='Joint3'
)

def draw_robot(theta1, theta2, theta3):

    # Convert degrees to radians
    t1 = np.radians(theta1)
    t2 = np.radians(theta2)
    t3 = np.radians(theta3)

    # Joint 1 position
    x1 = L1 * np.cos(t1)
    y1 = L1 * np.sin(t1)

    # Joint 2 position
    x2 = x1 + L2 * np.cos(t1 + t2)
    y2 = y1 + L2 * np.sin(t1 + t2)

    # End Effector position
    x3 = x2 + L3 * np.cos(t1 + t2 + t3)
    y3 = y2 + L3 * np.sin(t1 + t2 + t3)

    # Create Plot
    plt.figure(figsize=(7,7))

    # Link 1
    plt.plot(
        [0, x1],
        [0, y1],
        marker='o',
        linewidth=4
    )

    # Link 2
    plt.plot(
        [x1, x2],
        [y1, y2],
        marker='o',
        linewidth=4
    )

    # Link 3
    plt.plot(
        [x2, x3],
        [y2, y3],
        marker='o',
        linewidth=4
    )

    plt.xlim(-10, 10)
    plt.ylim(-10, 10)

    plt.grid()

    plt.title(
        f"3 DOF Robot Arm | End Effector = ({x3:.2f}, {y3:.2f})"
    )

    plt.show()


widgets.interactive(
    draw_robot,
    theta1=joint1,
    theta2=joint2,
    theta3=joint3
)
```

```python
def robot_info(theta1, theta2, theta3):

    print("===================================")
    print("          ROBOT INFORMATION")
    print("===================================")

    print("Robot Type       : 3-DOF RRR Robot")
    print("Number of Links  : 3")
    print("Number of Joints : 3")
    print("Degrees of Freedom: 3")

    print("\nJoint Information:")
    print(f"Joint 1 (R) : {theta1}°")
    print(f"Joint 2 (R) : {theta2}°")
    print(f"Joint 3 (R) : {theta3}°")

    print("\nLink Lengths:")
    print(f"Link 1 : {L1}")
    print(f"Link 2 : {L2}")
    print(f"Link 3 : {L3}")

    print("===================================")


widgets.interactive(
    robot_info,
    theta1=joint1,
    theta2=joint2,
    theta3=joint3
)
```
