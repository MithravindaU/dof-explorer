```python
import numpy as np
import matplotlib.pyplot as plt

# Link lengths
L1 = 4
L2 = 3
L3 = 2


# Joint angles (degrees)
theta1 = float(input("Enter Joint 1 Angle: "))
theta2 = float(input("Enter Joint 2 Angle: "))
theta3 = float(input("Enter Joint 3 Angle: "))

# Convert to radians
t1 = np.radians(theta1)
t2 = np.radians(theta2)
t3 = np.radians(theta3)

# Joint 1 position
x1 = L1 * np.cos(t1)
y1 = L1 * np.sin(t1)

x2 = x1 + L2 * np.cos(t1 + t2)
y2 = y1 + L2 * np.sin(t1 + t2)

x3 = x2 + L3 * np.cos(t1 + t2 + t3)
y3 = y2 + L3 * np.sin(t1 + t2 + t3)

plt.figure(figsize=(6,6))

plt.plot([0,x1],[0,y1],marker='o',linewidth=4)
plt.plot([x1,x2],[y1,y2],marker='o',linewidth=4)
plt.plot([x2,x3],[y2,y3], marker='o', linewidth=4)

plt.xlim(-8,8)
plt.ylim(-8,8)

plt.grid()
plt.title("2-Link Robot Arm")

plt.show()
```
